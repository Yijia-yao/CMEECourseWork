# Exercise 2 — Allometric scaling (power laws)

Plant traits often follow **power-law (allometric) scaling**. Suppose:

- **Leaf area** \(L\) is proportional to **stem diameter** \(D\):
  \[
  L = k D^{1.84}
  \]

- **Spongy mesophyll volume fraction** \(S\) is proportional to **leaf thickness** \(T\):
  \[
  S = c T^{-0.49}
  \]

## Questions

1. Sketch (qualitatively) how \(L\) changes with \(D\), and how \(S\) changes with \(T\).

2. Using **log–log axes**, explain how you could estimate the exponent from data.

3. For \(k = 1\), compute \(L\) when \(D = 1, 2, 4\).
   For \(c = 1\), compute \(S\) when \(T = 1, 2, 4\).

---


```python
import numpy as np
import matplotlib.pyplot as plt

# Leaf area L scales with stem diameter D:
# L = k * D^1.84
# Spongy mesophyll volume fraction S scales with leaf thickness T:
# S = c * T^(-0.49)

a = 1.84     
b = -0.49    
k = 1.0
c = 1.0

# Q1
# Q1(a): How does L change with D?
# Since the exponent 1.84 > 1, leaf area L increases more than linearly with D.
# This means:
# L increases as D increases
# the curve is convex upward (slope increases with D)
# small increases in D lead to increasingly large increases in L

# Q1(b): How does S change with T?
# Since the exponent -0.49 is negative, S decreases as T increases.
# This means:
# S is high for small T
# S decreases with increasing T
# the curve drops quickly at small T and flattens at larger T

D = np.linspace(0.5, 5, 400)
T = np.linspace(0.5, 5, 400)

L = k * D**a
S = c * T**b

plt.figure()
plt.plot(D, L)
plt.xlabel("Stem diameter D")
plt.ylabel("Leaf area L")
plt.title("Q1: L increases nonlinearly with D (L = D^1.84)")
plt.show()

plt.figure()
plt.plot(T, S)
plt.xlabel("Leaf thickness T")
plt.ylabel("Spongy mesophyll fraction S")
plt.title("Q1: S decreases with T (S = T^-0.49)")
plt.show()




# Q2
# A general power-law relationship is:
#   Y = A * X^m

# Taking logarithms:
#   log(Y) = log(A) + m * log(X)

# Therefore:
# plotting log(Y) vs log(X) gives a straight line
# the slope of this line equals the exponent m
# the intercept equals log(A)

# Plot the same relationships on log–log axes to illustrate this
plt.figure()
plt.loglog(D, L)
plt.xlabel("log(D)")
plt.ylabel("log(L)")
plt.title("Q2: L vs D on log–log axes (slope = 1.84)")
plt.show()

plt.figure()
plt.loglog(T, S)
plt.xlabel("log(T)")
plt.ylabel("log(S)")
plt.title("Q2: S vs T on log–log axes (slope = -0.49)")
plt.show()




# Q3
# Compute L for D = 1, 2, 4 when k = 1
D_vals = np.array([1, 2, 4], dtype=float)
L_vals = k * D_vals**a

# Compute S for T = 1, 2, 4 when c = 1
T_vals = np.array([1, 2, 4], dtype=float)
S_vals = c * T_vals**b

print("Q3: Numerical results\n")

print("Leaf area L = D^1.84 (k = 1):")
for D_i, L_i in zip(D_vals, L_vals):
    print(f"  D = {D_i:.0f} -> L = {L_i:.4f}")

print("\nSpongy mesophyll fraction S = T^-0.49 (c = 1):")
for T_i, S_i in zip(T_vals, S_vals):
    print(f"  T = {T_i:.0f} -> S = {S_i:.4f}")
```

# Exercise 4 — Oscillations and seasonality

A zooplankton abundance time series shows seasonal oscillations. Model it as:

$$
Z(t) = Z_0 + A \cos(\omega t + \phi)
$$

## Questions

1. Describe the role of $Z_0$, $A$, $\omega$, and $\phi$.
2. If the period is 12 months, what is $\omega$ (in radians per month)?
3. Plot $Z(t)$ over 24 months for $Z_0 = 50$, $A = 20$, $\phi = 0$, and compare $\omega$ vs $2\omega$ (higher frequency).


```python
import numpy as np
import matplotlib.pyplot as plt


# Q1
# Z0  : baseline means the abundance around which oscillations occur
# A   : amplitude of oscillation (distance from mean to peak)
# omega (ω): angular frequency (radians per month),
#            controls how fast the oscillation occurs
# phi (φ): phase shift, determines the timing of peaks/troughs
#
# The period P and omega are related by:
#   P = 2π / ω




# Q2
P = 12 
omega = 2 * np.pi / P

print("Q2: Angular frequency")
print(f"Period = {P} months")
print(f"omega = 2π / P = {omega:.6f} radians/month")




# Q3
Z0 = 50
A = 20
phi = 0

t = np.linspace(0, 24, 1000)  

Z_omega = Z0 + A * np.cos(omega * t + phi)
Z_2omega = Z0 + A * np.cos(2 * omega * t + phi)

# Using 2*omega doubles the frequency and halves the period,
# producing twice as many oscillations in the same time interval.

plt.figure()
plt.plot(t, Z_omega, label="omega (12-month period)")
plt.plot(t, Z_2omega, label="2*omega (6-month period)")
plt.xlabel("t (months)")
plt.ylabel("Z(t)")
plt.title("Seasonal oscillations: omega vs 2*omega")
plt.legend()
plt.show()
```

# Linear algebra-Exercise 3 — Inverses and solving linear systems (metabolic fluxes)

Suppose two metabolic fluxes $x, y$ satisfy:

$$
\begin{cases}
3x - 7y = 4 \\
x + 7y = 10
\end{cases}
$$

## Questions

1. Write this system in the form $A\mathbf{z}=\mathbf{b}$.
2. Compute $A^{-1}$ and hence $\mathbf{z}=A^{-1}\mathbf{b}$.
3. Verify numerically.


```python
import sympy as sp
import numpy as np

# Q1
# Let z = [x, y]^T
# Then:
#   [ 3  -7 ] [x] = [ 4 ]
#   [ 1   7 ] [y]   [10]

A = sp.Matrix([[3, -7],
               [1,  7]])
b = sp.Matrix([4, 10])

print("Q1: Matrix form A z = b")
print("A ="); sp.pprint(A)
print("b ="); sp.pprint(b)




# Q2
A_inv = A.inv()
z = A_inv * b

print("\nQ2: Inverse matrix and solution")
print("A^{-1} ="); sp.pprint(A_inv)
print("z = A^{-1} b ="); sp.pprint(z)

x_sol, y_sol = z
print(f"\nSolution: x = {x_sol}, y = {y_sol}")




# Q3
check = A * z
print("\nQ3: Verification (A*z should equal b)")
print("A*z ="); sp.pprint(check)
print("b   ="); sp.pprint(b)
print("Verified:", sp.simplify(check - b) == sp.zeros(2, 1))

# Numerical verification using numpy
A_np = np.array(A.tolist(), dtype=float)
b_np = np.array(b.tolist(), dtype=float)
z_np = np.linalg.solve(A_np, b_np)

print("\nNumeric check using numpy:")
print("z =", z_np)
```
